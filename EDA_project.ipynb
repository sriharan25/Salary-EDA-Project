```python
df= df.drop_duplicates()
```

```python
df["experience_level"].value_counts().plot(kind="bar")
plt.title("Experience Level Distribution")
plt.xlabel("Experience Level")
plt.ylabel("Count")
plt.show()                                           
```

```python
df["employment_type"].value_counts().plot(kind="pie",
autopct="%1.1f%%")
plt.title("Employment Type")
plt.show()
```

```python
df["experience_level"].value_counts().plot(kind="bar")

plt.title("Experience Level Distribution")
plt.xlabel("Experience Level")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df["salary_in_usd"], bins=20)

plt.title("Salary Distribution")
plt.xlabel("Salary in USD")
plt.ylabel("Employees")
plt.show()
```

```python
sns.countplot(x="company_size", data=df)

plt.title("Company Size")
plt.show()
```

```python
sns.countplot(x="remote_ratio", data=df)

plt.title("Remote Work Ratio")
plt.show()
```

```python
sns.countplot(x="work_year", data=df)

plt.title("Jobs by Year")
plt.show()
```

```python
top_jobs = df["job_title"].value_counts().head(10)

top_jobs.plot(kind="bar")

plt.title("Top 10 Job Titles")
plt.show()
```

Conclusion

1. Most employees are Full-Time.
2. Salaries vary significantly.
3. Medium-sized companies hire the most employees.
4. Remote jobs are common.
5. Data Scientist and Data Engineer are among the most common job titles.
